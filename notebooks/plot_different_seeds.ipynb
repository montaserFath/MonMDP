```python
import numpy as np
import matplotlib.pylab as plt
```

```python
n_seed = 10
grid_size = (9, 9)
n_checkpoints = 50
gamma = 0.99
episodes_per_checkpoint = 5e4
main_dir = "../cc_general_models/9_9/Penalty/reward_model/env_0.0/eps_decay/window_5/q_lr_0.001/reward_lr_0.001/"
```

```python
fig = plt.figure(figsize=(6, 4))
x_axis = episodes_per_checkpoint * np.arange(n_checkpoints)
for seed in range(n_seed):
    eval_rewards = np.load(main_dir + "/eval_joint_reward_{}.npy".format(seed))
    plt.plot(x_axis, eval_rewards, lw=3, label="seed={}".format(seed))
    plt.scatter(x_axis, eval_rewards, marker="o", s=1)
plt.hlines(gamma** 23, 0, x_axis[-1], color="r", lw=3, linestyle="--", label="Optimal")
plt.xlabel("Training Timesteps")
plt.ylabel("Eval Joint Episode Reward")
plt.grid(axis="y")
plt.legend()
# plt.ylim(-0.01, 0.85)
plt.tight_layout()
# fig.savefig(main_dir + "/all_seeds_eval_joint_reward.png", dpi=300)
```
