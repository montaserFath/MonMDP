```python
import numpy as np
```

```python
n_steps = "5k"
n_seed = 30
env_random = 0.1
q_lr = 0.1
r_lr = 0.1

eps = 1.0

lenght, rewards = np.zeros(n_seed), np.zeros(n_seed)
for seed in range(n_seed):
    log_dir = "../models_{}/9_9/Penalty/reward_model/env_{}/eps_{}/q_lr_{}/reward_lr_{}".format(n_steps, env_random, eps, q_lr, r_lr)
    traj = np.load(log_dir + "/trajectories_{}.npy".format(seed), allow_pickle=True)[()]
    gamma = 0.99
    ep_length = []
    ep_reward = []
    for key in traj.keys():
        ep_length.append(len(traj[key]["states"][:, 0]))
        ep_reward.append(np.sum(traj[key]["undiscounted_joint_reward"] * np.array([gamma**i for i in range(len(traj[key]["undiscounted_joint_reward"]))])))
    # print("episode length ", np.mean(ep_length), r"$\pm$", np.std(ep_length))
    # print("episode joint reward {:.3f} +- {:.3f}".format(np.mean(ep_reward), np.std(ep_reward)))
    lenght[seed], rewards[seed] = np.mean(ep_length), np.mean(ep_reward)
print("mean lenght {:.1f}, std = {:.2f}".format(np.mean(lenght), np.std(lenght)))
print("mean reward {:.3f}, std = {:.3f}".format(np.mean(rewards), np.std(rewards)))
```
